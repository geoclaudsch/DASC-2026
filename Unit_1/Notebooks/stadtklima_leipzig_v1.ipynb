## Gibt es einen Trend?

```python
# Nur gültige Temperaturwerte für die Trendberechnung
df_trend = df.dropna(subset=["JA_TT"]).copy()

# Jahreszahlen aus dem Datum extrahieren
x = df_trend["MESS_DATUM_BEGINN"].dt.year
y = df_trend["JA_TT"]

# Lineare Regression (Polynom 1. Grades: y = m*x + b)
m, b = np.polyfit(x, y, 1)

print(f"Temperaturänderung: {m:.3f} °C pro Jahr")
print(f"Temperaturänderung: {m*10:.2f} °C pro Jahrzehnt")
```

```python
plt.figure(figsize=(12, 5))

# Gemessene Jahresmitteltemperaturen
plt.plot(
    df["MESS_DATUM_BEGINN"],
    df["JA_TT"],
    label="Jahresmitteltemperatur",
    color="steelblue"
)

# Lineare Trendlinie
plt.plot(
    df_trend["MESS_DATUM_BEGINN"],
    m * x + b,
    label=f"Linearer Trend ({m*10:+.2f} °C/Jahrzehnt)",
    color="red",
    linewidth=2
)

plt.xlabel("Jahr")
plt.ylabel("Jahresmitteltemperatur [°C]")
plt.title("Jahresmitteltemperatur in Leipzig mit linearem Trend")

plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

## Alternativ: Daten via API einlesen

Manueller Download nicht notwendig, wenn ein Application Programming Interface - API vorhanden ist.
Wir brauchen Paket **requests**.

Wird von der datenbereitstellenden Stelle beschrieben (z.B. Swagger UI - grafische Benutzeroberfläche (UI) für Programmierschnittstellen (APIs)).

```python
import requests
import pandas as pd

# API-Endpunkt
url = "https://opendata.leipzig.de/api/3/action/datastore_search"

# ID unseres Datensatzes
resource_id = "62d8dd30-311d-4e06-89bd-38f2931cd8e8"

# Anfrage an die API
response = requests.get(
    url,
    params={
        "resource_id": resource_id,
        "limit": 1000
    },
    timeout=30
)

response.raise_for_status()

# JSON-Antwort auslesen
data = response.json()

print("Anfrage erfolgreich:", data["success"])
print("Anzahl Datensätze:", data["result"]["total"])
```

```python
df_api = pd.DataFrame(data["result"]["records"])
df_api.head()
```

```python
df_api.columns
```

```python
spaltennamen = [
    "STATIONS_ID",
    "MESS_DATUM_BEGINN",
    "MESS_DATUM_ENDE",
    "QN_4",
    "JA_N",
    "JA_TT",
    "JA_TX",
    "JA_TN",
    "JA_FK",
    "JA_SD_S",
    "JA_MX_FX",
    "JA_MX_TX",
    "JA_MX_TN",
    "QN_6",
    "JA_RR",
    "JA_MX_RS",
    "eor"
]
```

```python
# Technische CKAN-ID ausschließen
spalte = [c for c in df_api.columns if c != "_id"][0]

# Inhalte anhand des Semikolons aufteilen
df1 = df_api[spalte].str.split(";", expand=True)

# Spaltennamen zuweisen
df1.columns = spaltennamen

df1.head()
```

```python
print(df1.shape)
print(df1.columns)
```

```python
import numpy as np

df1["JA_TT"] = pd.to_numeric(df1["JA_TT"], errors="coerce")
df1["JA_TT"] = df1["JA_TT"].replace(-999, np.nan)

df1["MESS_DATUM_BEGINN"] = pd.to_datetime(
    df1["MESS_DATUM_BEGINN"].str.strip(),
    format="%Y%m%d"
)
```

```python
plt.figure(figsize=(12, 5))

plt.plot(df1["MESS_DATUM_BEGINN"], df1["JA_TT"])

plt.xlabel("Jahr")
plt.ylabel("Jahresmitteltemperatur [°C]")
plt.title("Jahresmitteltemperatur in Leipzig")

plt.grid(alpha=0.3)
plt.show()
```
